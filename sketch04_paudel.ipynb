```python
import pandas as pd  # importing pandas library
base_url = "https://raw.githubusercontent.com/pjbitterman/KSU_x9076/main/data" #reference to data directory

precip = pd.read_csv(f"{base_url}/part1/ohio_precip_2025.csv") #reading a csv file
print(f"First Five Rows of DataFrame:\n")
print(precip.head()) # gives first five rows of dataframe
print(f"\n Statistics:\n")
print(precip.describe()) #Describes data - statistics of data

```

```python
subset_df = precip[['name','precip_mm_2025']] #new data frame with just name and precip_mm_2025 columns
sorted = subset_df.sort_values(by='precip_mm_2025',ascending=False) # sort rows by precipitation value on descending order(puts highest value on top)
sorted.head()
```

```python
print(sorted.tail()) # print last five rows of dataframe - stations with least precipitation
```

The mean annual precipitation was 1008.86mm, and the wettest station was CINCINNATI MUNI AP LUNKEN FLD, with precipitation of 1495.7mm. This does not surprise me, as a difference of 486.84 mm from the mean is not unusual over a year. What actually surprises me is the difference between the wettest station, CINCINNATI MUNI AP LUNKEN FLD, with 1495.7 mm precipitation, and the driest station, ELYRIA LORAIN CO AP, with just 350.9 mm precipitation. This difference of 1144.8mm shows the large variation in annual precipitation across Ohio in 2025.
